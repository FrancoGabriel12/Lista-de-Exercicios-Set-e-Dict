## Exercício Conta Bancária

#### **1. Defina a classe `ContaBancaria`**
A classe `ContaBancaria` será responsável por representar uma conta bancária, com atributos e métodos que permitem gerenciar operações básicas, como depósito, saque e consulta de saldo.

#### **2. Atributos da Classe**
Ao instanciar a classe, os seguintes atributos devem ser fornecidos:
- **Número da conta**: Identificador único da conta.
- **Nome da pessoa**: Nome do titular da conta.
- **Balanço inicial**: Saldo inicial da conta em reais.
- **Tipo de conta**: Pode ser "corrente", "poupança" ou "investimento".

#### **3. Métodos da Classe**
A classe deve ter três métodos de instância:
1. **`deposito(quantidade)`**:
   - Adiciona um valor ao saldo da conta.
   - Valida se o valor do depósito é positivo.
     - Retorne uma mensagem de erro se o valor for negativo.
     - Retorne uma mensagem de erro se o valor for zero.
     - Retorne uma mensagem de erro se o valor for inválido.
2. **`saque(quantidade)`**:
   - Subtrai um valor do saldo da conta.
   - Valida se o valor do saque é positivo e se há saldo suficiente.
     - Retorne uma mensagem de erro se o valor for maior que o saldo disponível.
3. **`saldo()`**:
   - Retorna o saldo atual da conta.

#### **4. Teste das Operações**
Um exemplo de teste deve ser fornecido para demonstrar o funcionamento da classe. O teste deve:
- Criar três instâncias da classe `ContaBancaria`.
- Realizar operações de depósito e saque.
- Exibir o saldo após as operações.

#### **5. Adicionando Métodos de Classe e Estáticos**
- Permita o rastreamento do número de instâncias criadas da classe `ContaBancaria`
- Mostre na tela uma mensagem toda vez que uma instância da classe é criada
- Mostre na tela a quantidade de contas criadas
- Adicione um método estático que retorne o nome do banco `POD Banking`.

#### **6. Sobrecarga e Métodos Especiais**
- Adicione um método especial `__str__` que retorne uma representação em string da conta bancária, incluindo o número da conta, nome do titular e saldo.
- Adicione um método especial `__eq__` que permita comparar duas contas bancárias com base apenas no número da conta.
- Permita o uso dos operadores `+` e `+=` para somar os saldos de duas contas bancárias. 
  - O operador `+` deve retornar uma nova instância da classe `ContaBancaria` com o saldo combinado.
  - O operador `+=` deve atualizar o saldo da conta original com o saldo da outra conta. Você precisará zerar o saldo da conta que foi somada.
```

````python
class ContaBancaria:
    """Representa uma conta bancária simples."""

    numero_de_contas = 0          # atributo de classe: contador de instâncias criadas
    __nome_banco = "POD Banking"  # atributo de classe privado

    def __init__(self, numero_conta, nome, saldo_inicial, tipo_conta):
        self.numero_conta = numero_conta
        self.nome = nome
        self.saldo_inicial = saldo_inicial
        self.tipo_conta = tipo_conta

        ContaBancaria.numero_de_contas += 1
        print(f"Conta criada para {self.nome} (nº {self.numero_conta}). "
              f"Total de contas: {ContaBancaria.numero_de_contas}")

    def deposito(self, quantidade):
        if not isinstance(quantidade, (int, float)) or isinstance(quantidade, bool):
            return "Erro: valor de depósito inválido."
        if quantidade < 0:
            return "Erro: não é possível depositar um valor negativo."
        if quantidade == 0:
            return "Erro: não é possível depositar um valor igual a zero."
        self.saldo_inicial += quantidade
        return f"Depósito de R$ {quantidade:.2f} realizado com sucesso."

    def saque(self, quantidade):
        if not isinstance(quantidade, (int, float)) or isinstance(quantidade, bool):
            return "Erro: valor de saque inválido."
        if quantidade < 0:
            return "Erro: não é possível sacar um valor negativo."
        if quantidade > self.saldo_inicial:
            return "Erro: saldo insuficiente para realizar o saque."
        self.saldo_inicial -= quantidade
        return f"Saque de R$ {quantidade:.2f} realizado com sucesso."

    def saldo(self):
        return self.saldo_inicial

    @staticmethod
    def nome_banco():
        return ContaBancaria.__nome_banco

    @classmethod
    def total_de_contas(cls):
        return cls.numero_de_contas

    def __str__(self):
        return f"Conta {self.numero_conta} - Titular: {self.nome} - Saldo: R$ {self.saldo_inicial:.2f}"

    def __eq__(self, other):
        if not isinstance(other, ContaBancaria):
            return NotImplemented
        return self.numero_conta == other.numero_conta

    def __add__(self, other):
        if not isinstance(other, ContaBancaria):
            return NotImplemented
        novo_saldo = self.saldo_inicial + other.saldo_inicial
        return ContaBancaria(
            f"{self.numero_conta}+{other.numero_conta}",
            f"{self.nome} & {other.nome}",
            novo_saldo,
            self.tipo_conta,
        )

    def __iadd__(self, other):
        if not isinstance(other, ContaBancaria):
            return NotImplemented
        self.saldo_inicial += other.saldo_inicial
        other.saldo_inicial = 0
        return self


# ---- Teste das operações ----
conta1 = ContaBancaria(1, "Ana", 1000, "corrente")
conta2 = ContaBancaria(2, "Bruno", 500, "poupança")
conta3 = ContaBancaria(3, "Carla", 2000, "investimento")

print(conta1.deposito(200))
print(conta1.saque(150))
print(conta2.deposito(-50))
print(conta2.saque(10000))
print(conta3.deposito(0))
print(conta3.deposito("abc"))

print("Saldo conta1:", conta1.saldo())
print("Saldo conta2:", conta2.saldo())
print("Saldo conta3:", conta3.saldo())

print("Total de contas criadas:", ContaBancaria.total_de_contas())
print("Banco:", ContaBancaria.nome_banco())

print(conta1)
print(conta2)

conta1_copia = ContaBancaria(1, "Ana Cópia", 999, "corrente")
print("conta1 == conta1_copia:", conta1 == conta1_copia)
print("conta1 == conta2:", conta1 == conta2)

conta_soma = conta1 + conta2
print("Nova conta (soma):", conta_soma)

print("Antes do +=: conta3 =", conta3.saldo(), "| conta_soma =", conta_soma.saldo())
conta3 += conta_soma
print("Depois do +=: conta3 =", conta3.saldo(), "| conta_soma =", conta_soma.saldo())
````
