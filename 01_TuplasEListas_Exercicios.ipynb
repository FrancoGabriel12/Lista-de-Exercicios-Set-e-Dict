```python

#1. Intercalação de Listas:
#Crie uma função intercalar_listas(lista1, lista2) que receba duas listas e
#retorne uma nova lista onde os elementos de lista1 e lista2 sejam intercalados.
#Caso uma lista tenha mais elementos que a outra, os elementos restantes devem ser
#adicionados ao final.
#lista1 = [1, 3, 5]
#lista2 = [2, 4, 6, 8, 10]
#print(intercalar_listas(lista1, lista2))
# Saída esperada: [1, 2, 3, 4, 5, 6, 8, 10]
```

```python
# 1. Intercalação de Listas
def intercalar_listas(lista1, lista2):
    resultado = []
    for i in range(max(len(lista1), len(lista2))):
        if i < len(lista1):
            resultado.append(lista1[i])
        if i < len(lista2):
            resultado.append(lista2[i])
    return resultado

# Testando a função
lista1 = [1, 3, 5]
lista2 = [2, 4, 6, 8, 10]
print(intercalar_listas(lista1, lista2))
# Saída esperada: [1, 2, 3, 4,
```

```python
#Tuplas de pares ordenados:
#Crie uma função criar_pares(lista1, lista2) que receba duas listas e retorne
#uma lista de tuplas onde cada tupla contenha um elemento de lista1 e o respectivo
#elemento de lista2 na mesma posição. Caso uma lista seja maior, os elementos
#extras devem ser ignorados.

#lista1 = ['a', 'b', 'c']
#lista2 = [1, 2, 3, 4]
#print(criar_pares(lista1, lista2))
# Saída esperada: [('a', 1), ('b', 2), ('c', 3)]

```

```python
#2. Tuplas de pares ordenados
def criar_pares(lista1, lista2):
    return list(zip(lista1, lista2)) # A função zip combina os elementos das duas listas em tuplas
# Testando a função
lista1 = ['a', 'b', 'c']
lista2 = [1, 2, 3, 4]
print(criar_pares(lista1, lista2))

```

```python
# Remoção de duplicatas:
#Implemente uma função remover_duplicatas(lista) que receba uma lista e retorne
#uma nova lista contendo os mesmos elementos, mas sem duplicatas. A ordem dos
#elementos deve ser mantida.
#lista = [1, 2, 2, 3, 4, 4, 5]
#print(remover_duplicatas(lista))
# Saída esperada: [1, 2, 3, 4, 5]
```

```python
# 3. Remoção de duplicatas
def remover_duplicatas(lista):
    nova_lista = []
    for item in lista: # Itera sobre cada item na lista original
        if item not in nova_lista: # Verifica se o item já está na nova lista
            nova_lista.append(item) # Se não estiver, adiciona o item à nova lista
    return nova_lista # Retorna a nova lista sem duplicatas

# Testando a função
lista = [1, 2, 2, 3, 4, 4, 5]
print(remover_duplicatas(lista))
```

```python
#Filtrando tuplas por critérios:
#Crie uma função filtrar_tuplas(lista_tuplas, limite) que receba uma lista de
#tuplas onde cada tupla contém um nome e uma idade, e retorne apenas as tuplas onde
#a idade seja maior que um determinado limite.
#lista_tuplas = [('Ana', 23), ('João', 19), ('Maria', 30), ('Pedro', 17)]
#print(filtrar_tuplas(lista_tuplas, 20))
# Saída esperada: [('Ana', 23), ('Maria', 30)]
```

```python
# 4. Filtrando tuplas por critérios
def filtrar_tuplas(lista_tuplas, limite):
    for resultado in lista_tuplas: # Itera sobre cada tupla na lista de tuplas
        if resultado[1] > limite: # Verifica se a idade (segundo elemento da tupla) é maior que o limite
            print(resultado) # Se for, imprime a tupla
# Testando a função
lista_tuplas = [('Ana', 23), ('João', 19), ('Maria', 30), ('Pedro', 17)]
filtrar_tuplas(lista_tuplas, 20)
```

```python
# Soma de elementos pares
#Escreva uma função soma_pares(lista) que receba uma lista de números inteiros e
#retorne a soma dos números pares.
#lista = [1, 2, 3, 4, 5, 6]
#print(soma_pares(lista))
# Saída esperada: 12
```

```python
# 5. Soma de elementos pares
def soma_pares(lista):
    soma = 0
    for numero in lista: # Itera sobre cada número na lista
        if numero % 2 == 0: # Verifica se o número é par
            soma += numero # Se for, adiciona o número à soma
    return soma # Retorna a soma dos números pares
# Testando a função
lista = [1, 2, 3, 4, 5, 6]
print(soma_pares(lista))
```

```python
#Inverter Tupla:
#Crie uma função inverter_tupla(tupla) que receba uma tupla e retorne uma nova
#tupla com os elementos na ordem inversa.
#tupla = (1, 2, 3, 4, 5)
#print(inverter_tupla(tupla))
# Saída esperada: (5, 4, 3, 2, 1)
```

```python
# 6. Inverter Tupla
def inverter_tupla(tupla):
    return tupla[::-1] # Retorna a tupla invertida usando slicing
# Testando a função
tupla = (1, 2, 3, 4, 5)
print(inverter_tupla(tupla))
```

```python
# Produto cartesiano de listas:
#Escreva uma função produto_cartesiano(lista1, lista2) que retorne o produto
#cartesiano de duas listas, ou seja, uma lista de tuplas com todas as combinações
#possíveis de elementos de lista1 com lista2 .
#lista1 = [1, 2]
#lista2 = ['a', 'b']
#print(produto_cartesiano(lista1, lista2))
# Saída esperada: [(1, 'a'), (1, 'b'), (2, 'a'), (2, 'b')]
```

```python
# 7. Produto cartesiano de listas
def produto_cartesiano(lista1, lista2):
    resultado = []
    for item1 in lista1: # Itera sobre cada item na primeira lista
        for item2 in lista2: # Para cada item da primeira lista, itera sobre cada item da segunda lista
            resultado.append((item1, item2)) # Adiciona a tupla (item1, item2) ao resultado
    return resultado # Retorna a lista de tuplas do produto cartesiano
# Testando a função
lista1 = [1, 2]
lista2 = ['a', 'b']
print(produto_cartesiano(lista1, lista2))
```

```python
# Concatenar listas de tuplas:
#Crie uma função concatenar_listas_tuplas(lista1, lista2) que recebe duas
#listas de tuplas e retorna uma nova lista onde as tuplas em posições correspondentes
#são concatenadas (os valores dentro das tuplas são unidos em uma nova tupla).
#lista1 = [(1, 2), (3, 4)]
#lista2 = [(5, 6), (7, 8)]
#print(concatenar_listas_tuplas(lista1, lista2))
# Saída esperada: [(1, 2, 5, 6), (3, 4, 7, 8)]
```

```python
# 8. Concatenar listas de tuplas
def concatenar_listas_tuplas(lista1, lista2):
    resultado = []
    for tupla1, tupla2 in zip(lista1, lista2): # Itera sobre as duas listas de tuplas simultaneamente usando zip, que combina os elementos correspondentes de ambas as listas em tuplas
        resultado.append(tupla1 + tupla2) # Concatena as tuplas e adiciona ao resultado
    return resultado # Retorna a nova lista de tuplas concatenadas
# Testando a função
lista1 = [(1, 2), (3, 4)]
lista2 = [(5, 6), (7, 8)]
print(concatenar_listas_tuplas(lista1, lista2))
```
