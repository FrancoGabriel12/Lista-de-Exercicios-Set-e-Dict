# Tipos de Dados Compostos em Python

## Conjuntos

```python
#Conjuntos em Python
#Conjuntos são uma coleção de dados não ordenada, mutável e sem elementos duplicados. Eles são definidos usando chaves {} e os elementos são separados por vírgula.

conjunto = {1, 2, 3, 4, 5} # exemplo de conjunto
print(conjunto)
```

## Dicionários

```python
#Dicionários em Python
#Dicionários são uma coleção de dados mutável, não ordenada e indexada. Eles são definidos usando chaves {} e os elementos são separados por vírgula, onde cada elemento é composto por uma chave e um valor, separados por dois pontos :.
dicionario = {"nome": "João", "idade": 30, "cidade": "São Paulo"} # exemplo de dicionário
print(dicionario)
```

```python
#Outro exemplo de dicionário
dicionario2 = {"fruta": "maçã", "cor": "vermelha", "preço": 2.5} # exemplo de dicionário
print(dicionario2)

```
