# Questão 1: Verificar se um número é positivo, negativo ou zero
Solicite ao usuário um número e verifique se ele é positivo, negativo ou zero.

```python
numero = int(input("Digite um número inteiro: "))
if numero == 0:
    print("O número é zero.")
elif numero > 0:
    print("O número é positivo.")
else:
    print("O número é negativo.")
```

# Questão 4: Verificar se um número é primo
Solicite ao usuário um número e verifique se ele é primo.

```python
numero = int(input("Digite um número inteiro: "))
for i in range(2, numero // 2 + 1):
    if numero % i == 0:
        print(f"O número {numero} não é primo.")
        break
else:
    print(f"O número {numero} é primo.")
```

# Questão 7: Contar vogais em uma string
Solicite ao usuário uma string e conte quantas vogais ela contém.

```python
texto = input("Digite um texto: ").lower()
vogais = "aeiou"
contador = 0
for letra in texto:
    if letra in vogais:
        contador += 1
print(f"O texto contém {contador} vogais.")
```

# Questão 9: Verificar se uma string contém apenas números
Solicite ao usuário uma string e verifique se ela contém apenas números.

```python
entrada = input("Digite um texto: ")
if entrada.isdigit():
    print("O texto contém apenas números.")
else:
    print("O texto contém outros caracteres além de números.")
    
# alternativamente
for char in entrada:
    if not char.isdigit():
        print("O texto contém outros caracteres além de números.")
        break
else:
    print("O texto contém apenas números.")
```

# Questão 10: Calcular o fatorial de um número
Solicite ao usuário um número e calcule o fatorial desse número.

```python
numero = int(input("Digite um número inteiro não negativo: "))
fatorial = 1
for i in range(2, numero + 1):
    fatorial *= i
print(f"O fatorial de {numero} é {fatorial}.")
```

# Questão 13: Substituir vogais por '*' em uma string
Solicite ao usuário uma string e substitua todas as vogais por '*'.

```python
entrada = input("Digite um texto: ")
vogais = "aeiouAEIOU"
resultado = ""
for char in entrada:
    if char in vogais:
        resultado += "*"
    else:
        resultado += char
print("Texto modificado:", resultado)
```

# Questão 16: Verificar se um número é perfeito
Solicite ao usuário um número e verifique se ele é um número perfeito.

```python
numero = int(input("Digite um número inteiro positivo: "))
soma_divisores = 0
for i in range(1, numero):
    if numero % i == 0:
        soma_divisores += i
if soma_divisores == numero:
    print(f"O número {numero} é perfeito.")
else:
    print(f"O número {numero} não é perfeito.")
```

# Questão 17: Verificar se um número é um quadrado perfeito
Solicite ao usuário um número e verifique se ele é um quadrado perfeito.

```python
numero = int(input("Digite um número inteiro positivo: "))
raiz = int(numero**0.5)
if raiz * raiz == numero:
    print(f"O número {numero} é um quadrado perfeito.")
else:
    print(f"O número {numero} não é um quadrado perfeito.")
```

# Questão 18: Verificar se um número pertence à sequência de Fibonacci
Solicite ao usuário um número e verifique se ele é um número de Fibonacci.

```python
# check if a number belongs to Fibonacci sequence
n = int(input("Digite um número inteiro positivo: "))
a, b = 0, 1
while a < n:
    a, b = b, a + b
if a == n:
    print(f"O número {n} pertence à sequência de Fibonacci.")
else:
    print(f"O número {n} não pertence à sequência de Fibonacci.")
```
