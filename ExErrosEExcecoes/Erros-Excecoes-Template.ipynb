```python
class ErroIdadeInvalida(Exception):
    """"Classe de erro pra idade invalida"""
    def __init__(self, dado):
        self.__valor = dado
    def __str__(self):
        return "ErroIdadeInvalida: (" + str(self.__valor) + ") Idade deve ser um número inteiro entre 0 e 120"


#Utilizando Property
class Pessoa:
    """
    Exemplo de classe que armazena
    nome de pessoas e idade
    """

    def __init__(self, nome, idade): #nome do argumento do método
        self.nome = nome #atributo
        self.idade = idade

    def __str__(self):
        return self.nome + " possui " + str(self.idade) + " anos"
    
    @property
    def nome(self):
        print("getter")
        return self.__nome
    @nome.setter
    def nome(self, valor):
        self.__nome = valor
        print("setter")
    @nome.deleter
    def nome(self):
        print("deleter")
        del self.__nome
    
    def set_idade(self, valor):
        if isinstance(valor, int) and (valor > 0) and (valor < 120):
            self.__idade = valor
            print("setter")
        else:
            raise ErroIdadeInvalida(valor) #Lança a exceção personalizada ErroIdadeInvalida, passando o valor inválido como argumento. Isso permitirá que a mensagem de erro personalizada seja exibida quando a exceção for capturada.

    def get_idade(self):
        print("getter")
        return self.__idade
    
    idade = property(get_idade, set_idade, doc = "Propriedade para o atributo idade")
    
    def aniversario(self):
        self.idade+=1

try:
    NOME = input("Digite o nome da pessoa: ")
    IDADE = int(input("Digite a idade da pessoa: "))
    pessoa1 = Pessoa(NOME, IDADE)
except ErroIdadeInvalida:
    if IDADE < 0:
        IDADE = abs(IDADE)
    pessoa1 = Pessoa(NOME, IDADE)
print(pessoa1)
#del pessoa1
```

## Exceções Customizadas

```python
class ErroValorInvalido(Exception):
    """Exceção customizada para valores inválidos em operações bancárias."""
    def __init__(self, valor, mensagem="O valor informado é inválido"):
        self.valor = valor
        self.mensagem = mensagem
        super().__init__(f"{mensagem}: {valor}")

def sacar(saldo, valor):
    if valor <= 0:
        raise ErroValorInvalido(valor, "Valor de saque deve ser positivo")
    if valor > saldo:
        raise ErroValorInvalido(valor, "Saldo insuficiente para o saque")
    return saldo - valor

try:
    novo_saldo = sacar(100, -50)
except ErroValorInvalido as erro:
    print("Erro capturado:", erro)

try:
    novo_saldo = sacar(100, 500)
except ErroValorInvalido as erro:
    print("Erro capturado:", erro)

novo_saldo = sacar(100, 30)
print("Saque realizado com sucesso. Novo saldo:", novo_saldo)
```

## Exceções em cadeia

```python
class ErroDivisao(Exception):
    """Error de divisão"""
    def __init__(self, a, b):
        self.__a = a
        self.__b = b
    def __str__(self):
        return "ErroDivisao ( a=" + str(self.__a) + ", b=" + str(self.__b) + " )"

def divide(a, b):
    try:
        return a / b
    except Exception as e:
        raise ErroDivisao(a, b) from e #Lança a exceção personalizada ErroDivisao, passando os valores de a e b como argumentos. O "from e" é usado para indicar que a nova exceção está relacionada à exceção original capturada (e), permitindo que a cadeia de exceções seja mantida para fins de depuração.

res = divide(10, 0)
print(res)
```

# Exercício de Fixação: Conta Bancária (Cont-4)

1. Adicionar tratamento de exceção na classe `ContaCorrente` implementada no exercício anterior, onde não é permitido o saque e depósito de valores negativos.
    - Defina uma nova classe para tratamento de exceção chamada `ErroQuantidade`, que deve receber como parâmetro a conta e a mensagem de erro.
    - Atualize também os métodos de depósito e saque na classe `ContaBancária`  e `ContaCorrente` para que lancem a exceção  `ErroQuantidade`  se o valor fornecido for negativo.
    - Crie uma classe `ErroSaldo` para lidar com um possível saque abaixo ou acima do limite. É necessário mostrar a informação de qual conta gerou este erro.
    - Use blocos `try` e `except` para realizar os testes

```python
class ErroQuantidade(Exception):
    """Lançada quando um valor de depósito ou saque é inválido (negativo)."""
    def __init__(self, conta, mensagem):
        self.conta = conta
        self.mensagem = mensagem
        super().__init__(f"Conta {conta.numero_conta}: {mensagem}")

class ErroSaldo(Exception):
    """Lançada quando um saque ultrapassa o saldo (ou saldo + limite) disponível."""
    def __init__(self, conta, mensagem):
        self.conta = conta
        self.mensagem = mensagem
        super().__init__(f"Conta {conta.numero_conta} ({conta.nome}): {mensagem}")


class ContaBancaria:
    def __init__(self, numero_conta, nome, saldo_inicial=0):
        self.numero_conta = numero_conta
        self.nome = nome
        self.saldo = saldo_inicial

    def deposito(self, quantidade):
        if quantidade < 0:
            raise ErroQuantidade(self, "Não é possível depositar um valor negativo")
        self.saldo += quantidade
        return self.saldo

    def saque(self, quantidade):
        if quantidade < 0:
            raise ErroQuantidade(self, "Não é possível sacar um valor negativo")
        if quantidade > self.saldo:
            raise ErroSaldo(self, "Saldo insuficiente para realizar o saque")
        self.saldo -= quantidade
        return self.saldo

    def __str__(self):
        return f"Conta {self.numero_conta} - {self.nome} - Saldo: R$ {self.saldo:.2f}"


class ContaCorrente(ContaBancaria):
    def __init__(self, numero_conta, nome, saldo_inicial=0, limite=500):
        super().__init__(numero_conta, nome, saldo_inicial)
        self.limite = limite

    def saque(self, quantidade):
        if quantidade < 0:
            raise ErroQuantidade(self, "Não é possível sacar um valor negativo")
        if quantidade > (self.saldo + self.limite):
            raise ErroSaldo(
                self,
                f"Saque de R$ {quantidade:.2f} excede o saldo + limite "
                f"disponível (R$ {self.saldo + self.limite:.2f})",
            )
        self.saldo -= quantidade
        return self.saldo


# ---- Testes ----
conta_corrente = ContaCorrente(101, "Daniela", saldo_inicial=200, limite=300)

try:
    conta_corrente.deposito(-100)
except ErroQuantidade as erro:
    print("Erro:", erro)

try:
    conta_corrente.saque(-50)
except ErroQuantidade as erro:
    print("Erro:", erro)

try:
    conta_corrente.saque(1000)  # excede saldo + limite
except ErroSaldo as erro:
    print("Erro:", erro)

conta_corrente.deposito(100)
print(conta_corrente)

conta_corrente.saque(400)  # usa parte do limite disponível
print(conta_corrente)
```
